```python
from acl_anthology import Anthology
import pandas as pd
import re

# --- 1. Initialize the Anthology ---
# This will download the anthology data on the first run, which may take a few minutes.
# On subsequent runs, it will be much faster.
print("Initializing ACL Anthology...")
anthology = Anthology(datadir="acl-anthology/data")
print("Initialization complete.")
```

# Gathering data from ACL using API

```python
def extract_names_with_regex(author_string):
    #ONLY USE THIS FOR ACL
    pattern = r"first='(.*?)'.*?last='(.*?)'"
    
    matches = re.findall(pattern, author_string)
    
    # We then format each (first, last) tuple into a "First Last" string
    full_names = [f"{first} {last}" for first, last in matches]
    
    # Finally, we join them into a single comma-separated string
    return ", ".join(full_names)

def find_papers(query, anthology):
    papers_data = []
    for paper in anthology.papers():
        if int(paper.year) >= 2020:
            if (paper.title and query in paper.title.as_text().lower()) or \
                (paper.abstract and query in paper.abstract.as_text().lower()):
                    print(paper.title.as_text())
                    paper_dict = {
                    # Core IDs
                    #"id": paper.id,
                    #"full_id": paper.full_id,
                    #"bibkey": paper.bibkey,
                    "doi": paper.doi,
                    
                    # Content
                    "title": paper.title.as_text() if paper.title else None,
                    "abstract": paper.abstract.as_text() if paper.abstract else None,
                    "authors": extract_names_with_regex(str(paper.authors)),
                    #"editors": ", ".join([str(name) for name in paper.editors]),
                    
                    # Publication details
                    "year": paper.year,
                    "month": paper.month,
                    "publisher": paper.publisher,
                    "address": paper.address,
                    "journal": paper.journal,
                    "issue": paper.issue,
                    "pages": paper.pages,
                    "volume_id": paper.volume_id,
                    "collection_id": paper.collection_id,
                    
                    }
                    papers_data.append(paper_dict)
    return pd.DataFrame(papers_data)
```

```python
query = "controlled text generation"
find_papers(query, anthology).to_csv(f'{query}/acl_results.csv')
```

# DBLP Data gathering using API

```python
import requests

def get_json_response(query):
    response_json = requests.get(f"https://dblp.org/search/publ/api?q={[word+"$+" for word in query.split()]}&format=json")
    print(f'response status code:{response_json.status_code}')
    hits = (response_json.json()['result']['hits']['hit'])
    list_of_hits = []
    for hit in hits:
        if int(hit['info']['year']) >= 2020:
            list_of_hits.append(hit['info'])
    return pd.DataFrame(list_of_hits)
```

```python
def flatten_authors(authors_object):
    """Extracts author names and joins them into a single string."""
    # Check if the 'authors' key even exists
    if not isinstance(authors_object, dict) or 'author' not in authors_object:
        return "" # Return empty string if no author data

    author_data = authors_object['author']

    # Case 1: Multiple authors (it's a list of dicts)
    if isinstance(author_data, list):
        # Use a list comprehension to get the 'text' from each author dict
        names = [author['text'] for author in author_data]
    # Case 2: A single author (it's just one dict)
    else:
        names = [author_data['text']]

    # Join the list of names with a comma and space
    return ", ".join(names)
```

```python
query = "controlled text generation"
controlled_df = get_json_response(query)
controlled_df['authors'] = controlled_df['authors'].apply(flatten_authors)
```

```python
controlled_df.to_csv('controlled text generation/controlled_text_generation_dblp.csv')
```

```python
query = "constrained text generation"
constrained_df = get_json_response(query)
constrained_df['authors'] = constrained_df['authors'].apply(flatten_authors)
```

```python
constrained_df.to_csv('constrained text generation/constrained_text_generation_dblp.csv')
```

```python
query = "lexical constraints"
lexical_df = get_json_response(query)
lexical_df['authors'] = lexical_df['authors'].apply(flatten_authors)
```

```python
lexical_df.to_csv('lexical constraints/lexical_constraints_dblp.csv')
```

```python
import pandas as pd
import glob
import os

# Define the directories
dirs_to_process = [
    'constrained text generation',
    'controlled text generation',
    'lexical constraints'
]

# List to hold dataframes
all_dfs = []

# Loop through each directory
for directory in dirs_to_process:
    # Find all csv files in the directory
    csv_files = glob.glob(os.path.join(directory, '*.csv'))
    # Loop through the csv files
    for f in csv_files:
        # Read the csv file and append to the list of dataframes
        all_dfs.append(pd.read_csv(f))

# Concatenate all dataframes
merged_df = pd.concat(all_dfs, ignore_index=True)


# Remove papers with 'arxiv' in their DOI
#if 'doi' in merged_df.columns:
    # Store the number of rows before filtering
    #rows_before_arxiv_filter = len(merged_df)
    # Filter out papers where the DOI contains 'arxiv' (case-insensitive)
    #merged_df = merged_df[~merged_df['doi'].astype(str).str.contains('arxiv', case=False, na=False)]
    #rows_after_arxiv_filter = len(merged_df)
    #print(f"Removed {rows_before_arxiv_filter - rows_after_arxiv_filter} papers with 'arxiv' in DOI.")
    #print(f"Number of rows after removing arXiv papers: {rows_after_arxiv_filter}")

# De-duplicate based on title
# Before dropping, let's see how many duplicates we have
print(f"Number of rows before deduplication: {len(merged_df)}")
print(f"Number of duplicate titles: {merged_df['title'].duplicated().sum()}")

# Drop duplicates, keeping the first occurrence
final_df = merged_df.drop_duplicates(subset=['title'], keep='first')

print(f"Number of rows after deduplication: {len(final_df)}")


# Save to a new CSV file
final_df.to_csv('final_master_merge.csv', index=False)

print("Merged and deduplicated file saved as 'final_master_merge.csv'")
```

```python
import pandas as pd

# Load the datasets again in this cell to ensure it can be run independently
citation_df = pd.read_csv('citation_chained.csv')
master_df = pd.read_csv('final_master_merge.csv')

# Clean and standardize titles for comparison
# Using .astype(str) handles potential non-string data, .str.strip() removes whitespace, .str.lower() makes it case-insensitive, and removes punctuation
citation_titles_clean = citation_df['title'].astype(str).str.strip().str.lower().str.replace(r'[^\w\s]', '', regex=True)
master_titles_clean = master_df['title'].astype(str).str.strip().str.lower().str.replace(r'[^\w\s]', '', regex=True)

# Find which of the original citation_chained titles are in the master list
mask = citation_titles_clean.isin(master_titles_clean)
matching_titles = citation_df[mask]['title']

# Print the results
if not matching_titles.empty:
    print("The following papers from 'citation_chained.csv' are also in 'final_master_merge.csv':")
    # Pretty print the list of titles
    for title in matching_titles:
        print(f"- {title.strip()}")
else:
    print("No matching titles were found between the two files.")

```

```python
import pandas as pd
import numpy as np

# Load the two datasets
master_df = pd.read_csv('final_master_merge.csv')
citation_df = pd.read_csv('citation_chained.csv')

# Get the list of columns from the citation_chained.csv
citation_columns = citation_df.columns

# Find which columns are missing from the master dataframe
missing_cols = set(citation_columns) - set(master_df.columns)

# Add the missing columns to the master dataframe and initialize with blank values
for col in missing_cols:
    master_df[col] = np.nan

# Ensure the columns in master_df are in the same order as citation_df
# We'll use a list of columns that are present in both to avoid errors
ordered_cols = [col for col in citation_columns if col in master_df.columns]
master_df_reordered = master_df[ordered_cols]

# Overwrite the original final_master_merge.csv with the reformatted data
master_df_reordered.to_csv('final_master_merge.csv', index=False)

print("Reformatted 'final_master_merge.csv' to match the columns of 'citation_chained.csv'.")
print(f"Columns are now: {list(master_df_reordered.columns)}")

```
